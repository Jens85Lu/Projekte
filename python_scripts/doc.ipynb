Python Signalsimulation
----------------------------------
später: Arduino-Rechteck → RC-Tiefpass → Arduino misst → Python analysiert

* Rechtecksignal Arduino
* Pulsdauer 10 ms
* Wiederholrate 5 Hz

```python
import numpy as np
import matplotlib.pyplot as plt

# Rechtecksignal erzeugen mit 1 ms Periodendauer
def generate_square_wave(frequency, duration, sample_rate):
    t = np.linspace(0, duration, int(sample_rate * duration), endpoint=False)
    signal = 0.5 * (1 + np.sign(np.sin(2 * np.pi * frequency * t)))
    return t, signal

# Parameter definieren
frequency = 1  # Frequenz in Hz (1 ms Periodendauer)
duration = 10  # Dauer des Signals in Millisekunden
sample_rate = 100000  # Abtastrate in Hz

# Rechtecksignal erzeugen und plotten
t, signal = generate_square_wave(frequency, duration, sample_rate)
plt.figure(figsize=(10, 4))
plt.plot(t, signal, label="Rechtecksignal")
plt.title("Rechtecksignal mit 1 ms Periodendauer")
plt.xlabel("Zeit (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend(loc="upper right")
plt.show()
```

```python
T = 200 # entspricht 5Hz Wiederholrate
t_puls = 10 # Pulsdauer in ms
sample_rate =  20 # Abtastrate in Hz, also 20 Samples pro ms

def signal_generator(duration):
    t = np.linspace(0, duration, int(sample_rate * duration), endpoint=False)
    signal = np.zeros_like(t)
    for i in range(len(t)):
        if t[i] % T < t_puls:
            signal[i] = 1
    return t, signal

t, signal = signal_generator(1000) # 1 Sekunde Dauer
```

```python
plt.figure(figsize=(10, 4))
plt.plot(t, signal, label="Pulsfolge")
plt.title("Pulsfolge mit 5Hz Wiederholrate und 10ms Pulsdauer")
```
